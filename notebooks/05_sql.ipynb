```python
import sqlite3
import pandas as pd

con = sqlite3.connect("../data/zomato.db")
pd.read_csv("../data/zomato_ncr_clean.csv").to_sql(
    "restaurants", con, index=False, if_exists="replace")

with open("../sql/queries.sql", encoding="utf-8") as f:
    queries = [q.strip() for q in f.read().split(";") if q.strip()]

for q in queries:
    print(q.splitlines()[0])
    display(pd.read_sql(q, con))
```
